```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
master = pd.read_csv('../data/master_df.csv')
master.head()
```

1. How do socioeconomic factors (income, poverty rate, unemployment) correlate with opioid overdose mortality rates across U.S. counties?

```python
corr_cols = [
    "overdose_rate",
    "median_income",
    "poverty_rate",
    "unemployment_rate"
]

master[corr_cols].corr()["overdose_rate"].sort_values()
```

```python
plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    x=master["median_income"],
    y=master["poverty_rate"],
    s=master["overdose_rate"] * 5,  # Variable 3: Scale sizes to keep them visible
    c=master["unemployment_rate"],  # Variable 4: Mapped to a continuous color ramp
    cmap="viridis",  # Color palette (dark purple to bright yellow)
    alpha=0.7,  # Transparency to see overlapping points
    edgecolors="w",
)

plt.xlabel("Median Income ($)")
plt.ylabel("Poverty Rate")
plt.title("Correlation of Income, Poverty, Unemployment, and Overdoses", fontsize=14, pad=15)

# Add colorbar for Variable 4
cbar = plt.colorbar(scatter)
cbar.set_label("Unemployment Rate")

# Add a mock size legend for Variable 3
handles, labels = scatter.legend_elements(prop="sizes", alpha=0.6)
# Relabel the sizes to match original retention rates
size_legend = plt.legend(
    handles,
    ["20%", "40%", "60%", "80%", "100%"],
    loc="upper right",
    title="Overdose Rate",
)

plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
county_deaths = (
    master.groupby('county_state', as_index=False)['deaths']
      .sum()
      .sort_values('deaths', ascending=False)
)

county_deaths
```

```python
county_summary = master.groupby('county_state').agg({
    'deaths': 'sum',
    'population': 'sum'
}).reset_index().sort_values('deaths', ascending=False)
county_summary
```

```python
(county_summary['deaths']/county_summary['population']) * 100
```
