I am going to proceed to pull data from the U.S. Census Bureau (American Community Survey via API)

I want to include county_name, median_income, population, state code, county code.

```python
import pandas as pd
import requests
import matplotlib.pyplot as plt
```

```python
with open("./api_key.json", "r") as fi:
    api_key = json.load(fi)["api_key"]
```

```python
#for only the year 2020
API_KEY = api_key

url_base = (
    "https://api.census.gov/data/2020/acs/acs5"
    "?get=NAME,B19013_001E,B01003_001E"
    "&for=county:*"
    f"&key={API_KEY}"
)

r1 = requests.get(url_base)

print(r1.status_code)
print(r1.text[:300])  

base = r1.json()

df_base = pd.DataFrame(base[1:], columns=base[0])

url_subject = (
    "https://api.census.gov/data/2020/acs/acs5/subject"
    "?get=NAME,S1701_C03_001E,S2301_C04_001E"
    "&for=county:*"
    f"&key={API_KEY}"
)

r2 = requests.get(url_subject)

print(r2.status_code)
print(r2.text[:300])

subject = r2.json()

df_subject = pd.DataFrame(subject[1:], columns=subject[0])


# Rename

df_base.rename(columns={
    "NAME": "county_name",
    "B19013_001E": "median_income",
    "B01003_001E": "population"
}, inplace=True)

df_subject.rename(columns={
    "S1701_C03_001E": "poverty_rate",
    "S2301_C04_001E": "unemployment_rate"
}, inplace=True)


# Merge

df_2020 = df_base.merge(
    df_subject[
        ["state","county","poverty_rate","unemployment_rate"]
    ],
    on=["state","county"],
    how="left"
)

# numeric conversion
cols = [
    "median_income",
    "population",
    "poverty_rate",
    "unemployment_rate"
]

df_2020[cols] = df_2020[cols].apply(
    pd.to_numeric,
    errors="coerce"
)

df_2020.head()
```

Now I need to clean the columns so that the state is separate from county names. I also need to make the FIPS code for merging purposes. Then I will export it as a CSV.

```python
df_2020["fips"] = df_2020["state"] + df_2020["county"]
df_2020.head()
```

```python
df_2020.to_csv(
    "../data/raw/census_acs_2020.csv",
    index=False
)
```
